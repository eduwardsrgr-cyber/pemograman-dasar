```python
# Soal 2
print("=== ANALISIS ENERGI & CAPACITY FACTOR ===")
daya_kw = 450
jam = 8
erengi_aktual = 7200   # kWh
kapasitas_kw = 400

# Perhitungan Energi
energi = daya_kw * jam
print(f"Energi Harian Aktual  : {energi} kWh")

# Perhitungan Capacity Factor
capacity_factor = (erengi_aktual / (kapasitas_kw * 24)) * 100
print(f"Capacity Factor       : {capacity_factor:.2f}%")
```

```python
# Soal 6
suhu_list = [550, 540, 530, 530, 520]
tekanan_list = [10, 14, 13, 11, 13]

print("=== EVALUASI STATUS SISTEM ===\n")
print(f"{'No':<5} | {'Suhu (°C)':<10} | {'Tekanan (bar)':<14} | {'Status':<10}")
print("-" * 48)

for i in range(len(suhu_list)):
    suhu = suhu_list[i]
    tekanan = tekanan_list[i]

    if suhu >= 550 or tekanan >= 14:
        status = "TRIP"
    elif suhu > 520 and tekanan > 12:
        status = "WARNING"
    else:
        status = "NORMAL"

    print(f"{i+1:<5} | {suhu:<10} | {tekanan:<14} | {status:<10}")
```

```python
# Soal 10
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

for n, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
    except ValueError:
        print(f"Titik {n}: data tidak valid - pemeriksaan dihentikan")
        break

    if suhu > 70.0:
        print(f"Warning: Titik {n} suhu bearing tinggi!")

    jumlah_valid += 1

print(f"Jumlah data valid : {jumlah_valid}")
```

```python
# Soal bonus
BATAS_RENDAH = 50
BATAS_NORMAL = 100
BATAS_TINGGI = 150
BATAS_AMAN = 140
PENURUNAN = 10

# List untuk menyimpan data mentah (teks) beban per jam
data_mentah = ["32.5", "28.0", "45.2", "ERR", "78.9", "120.4",
               "160.8", "95.0", "110.3", "ERR", "55.5", "40.0"]

beban_valid = []
jam_valid = []
jumlah_rusak = 0

print("=== LAPORAN BEBAN LISTRIK HARIAN ===")

# Perulangan for: proses tiap data, jam dimulai dari 1
for jam, nilai in enumerate(data_mentah, start=1):

    try :
        beban = float(nilai)
    except ValueError:
        print(f"Jam {jam:>2}: data tidak valid ('{nilai}') - dilewati")
        jumlah_rusak += 1
        continue

    beban_valid.append(beban)
    jam_valid.append(jam)

    if beban < BATAS_RENDAH:
        status = "RENDAH"
    elif beban < BATAS_NORMAL:
        status = "NORMAL"
    elif beban < BATAS_TINGGI:
        status = "TINGGI"
    else:
        status = "KRITIS"

    print(f"Jam {jam:>2}: {beban:>6.1f} kW -> {status}")

    if status == "KRITIS":
        sisa = beban
        while sisa > BATAS_AMAN:
            sisa -= PENURUNAN
            print(f"         load shedding: beban turun ke {sisa:.1f} kW")
        print(f"         beban aman tercapai pada {sisa:.1f} kW")


print("\n=== RINGKASAN ===")
if len(beban_valid) > 0:
    total = sum(beban_valid)
    rata_rata = total / len(beban_valid)
    beban_puncak = max(beban_valid)
    jam_puncak = jam_valid[beban_valid.index(beban_puncak)]

    print(f"Data valid       : {len(beban_valid)} dari {len(data_mentah)} jam")
    print(f"Data rusak       : {jumlah_rusak}")
    print(f"Total konsumsi   : {total:.1f} kWh")
    print(f"Rata-rata beban  : {rata_rata:.1f} kW")
    print(f"Beban puncak     : {beban_puncak:.1f} kW (jam {jam_puncak})")
else:
    print("Tidak ada data valid untuk diproses.")
```
